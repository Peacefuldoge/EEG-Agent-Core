# EEG 可视化图表库：17 种图（通用）

演示任意 `(N,C,T)` EEG 数据的时域、频域、时频、空间与 QC 图。图表通过 `eeg_visualization.py` 绘制，与批量 HTML 报告使用相同函数。

````python
from pathlib import Path
import matplotlib.pyplot as plt
from IPython.display import display
from eeg_visualization import read_npz, plot_record, PLOTS

SOURCE = next(iter(sorted(Path('examples/demo_npz').glob('*.npz'))), None)
if SOURCE is None:
    print('请先运行 python examples/create_demo.py，或把 SOURCE 改成自己的 .npz 文件。')
else:
    print('Loading',SOURCE)
````

````python
if SOURCE is not None:
    rec = read_npz(SOURCE, max_windows=30)
    print(rec.dataset,rec.subject,rec.data.shape,rec.sfreq,'Hz')
    print('Available plots:', ', '.join(PLOTS))
````

````python
KINDS = ['traces','stacked_traces','psd','fft','spectrogram','bandpower',
         'bandpower_map','bandpower_time','fc_heatmap','fc_circle','fc_network',
         'fc_strength','rms','rms_time','channel_qc','fc_distribution','qc_summary']
if SOURCE is not None:
    for kind in KINDS:
        fig = plot_record(rec, kind, top_edges=20)
        display(fig)
        plt.close(fig)
````
